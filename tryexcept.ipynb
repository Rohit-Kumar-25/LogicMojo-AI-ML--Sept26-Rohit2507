```python
try:
    age=int(input("Enter the number"))
    if age=="":
        raise ValueError("No age passed")
    if age<0:
        raise ValueError("Age is neagtive")
    if age>=18:
        print("Eligible to vote")
    else:
        print("Not eligble to vote")
except ValueError as e:
    print(e)
```

```python
try:
    number=int(input("Enter the number"))
    if number<=0:
        raise ValueError("number is neagtive")
    else:
        if(number%2==0):
            print("Even number")
        else:
            print("Odd Number")
        
except ValueError as e:
    print(e)
```

```python
try:
    num1=int(input("Enter the number1"))
    num2=int(input("Enter the number2"))
    if num2==0:
        raise ZeroDivisionError("not able to divide")
    else:
        result=num1/num2
        print(result)

except ZeroDivisionError as e:
    print(e)
```
