```python
# Tuple containing model names and their accuracies
evaluation_results = (
    ("Model A", 0.85),
    ("Model B", 0.91),
    ("Model C", 0.88),
    ("Model D", 0.95)
)


bestmodel=evaluation_results[0][0]
highaccuracy=evaluation_results[0][1]
for model_name, accuracy in evaluation_results:
    print("Model Name:", model_name)
    print("Accuracy:", accuracy)
    if accuracy>highaccuracy:
        highaccuracy=accuracy
        bestmodel=model_name
        

# Print the best model
print("Model with the highest accuracy:", bestmodel)
print("Highest accuracy:",highaccuracy)
```
